## Using range() to Make a List of Numbers

```python
numbers = list(range(1,6))
print(numbers)
```

```python
# even numbers from 2 to 10
even_numbers = list(range(2, 12, 2)) # last index is exclusive
print(even_numbers)
```

```python
# square numbers
squares = []
for value in range(1,11):
    square = value ** 2
    squares.append(square)
print(squares)
```

```python
# square numbers
squares = []
for value in range(1,11):
    squares.append(value ** 2)
print(squares)
```

##  Simple Statistics with a List of Numbers

```python
digits = [1, 2, 3, 4, 5, 6, 7, 8, 9, 0]
print(min(digits))
print(max(digits))
print(sum(digits))
```

# List Comprehensions

-  A list comprehension allows you to generate 
this same list in just one line of code. 
- A list comprehension combines the 
for loop and the creation of new elements into one line, and automatically 
appends each new element

```python
squares = [value ** 2 for value in range(1,11)] # no colon is used here [experession for item in iterable]
print(squares)
```

## Slicing a List
- If you omit the first index in a slice, Python automatically starts your 
slice at the beginning of the list

```python
players = ['charles', 'martina', 'michael', 'florence', 'eli']
players[0:3]
```

```python
print(players[1:4])
print(players[:2])
print(players[1:])
```

```python
# to get last three elements from a list
print(players[-3:])
# skips how many elements
print(players[::2])
```

### looping through a slice

```python
players = ['charles', 'martina', 'michael', 'florence', 'eli']
print("Here are the first three players on my team")
for player in players[:3]:
    print(player.title())
```

## copying a list

 To copy a list, you can make a slice that includes the entire original list 
by omitting the first index and the second index ([:]). This tells Python to 
make a slice that starts at the first item and ends with the last item, produc
ing a copy of the entire list.

```python
my_foods = ['pizza', 'falafel', 'carrot cake']
friends_foods = my_foods[:]  # copying a list

print("My favorite foods are:")
print(my_foods)

print("\nMy friend's favorite foods are:")
print(friends_foods)

my_foods.append('cannoli')
friends_foods.append('ice cream')

print("My favorite foods are:")
print(my_foods)
print("\nMy friend's favorite foods are:")
print(friends_foods)
```

```python
# This doesn't work

my_foods = ['pizza', 'falafel', 'carrot cake']
friends_foods = my_foods  # this doesn't make a copy

my_foods.append('cannoli')
friends_foods.append('ice cream')

print("My favorite foods are:")
print(my_foods)
print("\nMy friend's favorite foods are:")
print(friends_foods)
```
