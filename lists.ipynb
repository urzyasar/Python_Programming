# Lists
- A list is a collection of items in a particular order. 
- Good idea to name the list in plural form. 
-  square brackets ([]) indicate a list, and individual elements 
in the list are separated by commas

```python
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
print(bicycles) # output will be in list representation
```

### Accessing elements in a list
- Lists are ordered collections
- elements can be accessed by telling the position
- Index Positions Start at 0, Not 1


```python
bicycles[0] # accessing first element
print(bicycles[0].title())
```

```python
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
print(bicycles[1])
print(bicycles[3])
print(bicycles[-1])  # accessing last element
print(bicycles[-2])  # accessing second last element
```

##  Modifying, Adding, and Removing Elements


```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)
motorcycles[0] = 'ducati' # modifying first element
print(motorcycles)
```

```python
# Appending Elements to the End of a List
motorcycles = ['honda', 'yamaha', 'suzuki']
motorcycles.append('ducati')
print(motorcycles)
```

```python
# Inserting Elements into a List
motorcycles = ['honda', 'yamaha', 'suzuki']
motorcycles.insert(0, 'ducati') # inserting at the beginning, can insert at any index
print(motorcycles)
```

### Removing Elements from a List
- if you know the position of the element, use del statement for removing
- using del, it is an in place deletion, no need to assign it back

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
del motorcycles[2]  # removing element at index 2
print(motorcycles)
```

###  Removing an Item Using the pop() Method
- It lets you use the value of item when it is removed from list
- it returns an element unlike del statement
- Popping Items from Any Position in a List

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)
popped_motorcycle = motorcycles.pop()  # removes last element and returns it
print(popped_motorcycle)
```

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
last_owned = motorcycles.pop()
print(f"The last motorcycle I owned was a {last_owned.title()}.")
```

```python
motorcycles = ['honda', 'yamaha', 'suzuki']
print(motorcycles)
first_owned = motorcycles.pop(0)  # removes element at index 0
print(first_owned)
print(f"The first motorcycle I owned was a {first_owned.title()}.")
print(motorcycles)
```

## when to use what?

-  If you’re unsure whether to use the del statement or the pop() method, 
here’s a simple way to decide: when you want to delete an item from a list 
and not use that item in any way, use the del statement; if you want to use an 
item as you remove it, use the pop() method

### Removing an Item by Value
- if won’t know the position of the value you want to remove 
from a list. If you only know the value of the item you want to remove, you 
can use the remove() method.
-  remove method returns nothing
- **note**: The remove() method deletes only the first occurrence of the value you specify. If there’s 
a possibility the value appears more than once in the list, you’ll need to use a loop 
to make sure all occurrences of the value are removed. 

```python
motorcycles = ['honda', 'yamaha', 'suzuki', 'ducati']
print(motorcycles)
motorcycles.remove('ducati')
print(motorcycles)
```

```python
motorcycles = ['honda', 'yamaha', 'suzuki', 'ducati']
print(motorcycles)
too_expensive = 'ducati'
test = motorcycles.remove(too_expensive)
print(test)  # None, as remove() does not return anything
print(motorcycles)
print(f"\nA {too_expensive.title()} is too expensive for me.")
```

## Organizing a list


### Sorting a List Permanently with the sort() Method
- inplace sort
- order of list is permanently changed



```python
cars = ['bmw', 'audi', 'toyota', 'subaru']
cars.sort()  # sorts the list in place
print(cars)
```

```python
# To sort in reverse order
cars.sort(reverse=True)
print(cars)
```

###  Sorting a List Temporarily with the sorted() Function
- The sorted() function lets you display your list 
in a particular order, but doesn’t affect the actual order of the list.

```python
cars = ['bmw', 'audi', 'toyota', 'subaru']
print("here is the original list:", cars)
print("\nhere is the sorted list:", sorted(cars))  # temporary sort
print("\nhere is the original list again:", cars)
```

```python
# reverse the order of a list
cars = ['bmw', 'audi', 'toyota', 'subaru']
cars = sorted(cars, reverse=True)
print(cars)
```

##  Printing a List in Reverse Order
- It simply reverses the order, not alphabetically
- again, it is a inplace reverse

```python
cars = ['bmw', 'audi', 'toyota', 'subaru']
print(cars)
cars.reverse()
print(cars)
```

###  Finding the Length of a List
- Python counts the items in a list starting with one, so you shouldn’t run into any 
off-by-one errors when determining the length of a list.

```python
cars = ['bmw', 'audi', 'toyota', 'subaru']
print(len(cars))
```

# working with lists

```python
magicians = ['alice', 'david', 'carolina']
for magician in magicians: # “For every magician in the list of magicians, print the magician’s name.” 
    print(magician)
```

```python
magicians = ['alice', 'david', 'carolina']
for magician in magicians:
    print(f"{magician.title()}, that was a great trick!")
    print(f"I can't wait to see your next trick, {magician.title()}.\n")
print("Thank you, everyone. That was a great magic show!")
```
